```python
import pandas as pd
employees={
    'empID':[101,102,103,104],
    'Name':['asha',None,"meena","sai"],
    'department':['sales','it','hr',None],
    'salary':[45000,60000,52000,89000]

}
I=[901,902,903,904]
df=pd.DataFrame(employees,index=I)
print(df.head(2))
print(df.tail(2))
print(df[['Name','salary']])
print(df['empID']==103)
print(df['salary']>50000)
print(df['department']=='it')
print(df[(df['Name']==None)|(df['department']==None)])
print(df.isna().sum())
print(df.fillna({'Name':'unknown','department':'not Assigned'}))
df['bonus'] = df['salary']*0.10
print(df)
df.rename(columns={'empID':'EmployeeID'})
print(df.drop('bonus',axis=1))
print(df['salary'].mean())
print(df['salary'].min())
print(df['salary'].max())
print(df['salary'].sum())
df=df['salary']=df['salary'].where(df['Name']!='sai',95000)
print(df)

emp_details={
    'empID':[101,105,103,107],
    'city':['hyd','che','ban','pune'],
    'experience':[2,4,3,5]
}
df1=pd.DataFrame(emp_details,index=I)
print(df1)
f=pd.merge(
    df,
    df1,
    on='empID',
    how='inner'
)
a = pd.merge(
    df,
    df1,
    on='empID',
    how='outer'
)
d=pd.merge(
    df,
    df1,
    on='empID',
    how='right'
)
g=pd.merge(
    df,
    df1,
    on='empID',
    how='left')
print(f)
print(d)
print(a)
print(g)
pd.concat([df,df1],axis=0,ignore_index=True)



```

```python
import pandas as pd
employees={
    'empID':[101,102,103,104],
    'Name':['asha',None,"meena","sai"],
    'department':['sales','it','hr',None],
    'salary':[45000,60000,52000,89000]

}

df=pd.DataFrame(employees)
print(df)
df['salary']=df['salary'].apply(
    lambda x:x*1.10
)
print(df)
dept={'sales':'Sales','it':'IT','hr':'HR'}
df['department']=df['department'].map(dept)
print(df)
```
