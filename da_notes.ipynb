```python
import pandas as pd
days=['monday','tuesday','wednesday','thursday','friday','saturday']
price=[5000,7000,6000,9000,8000,4000]
s=pd.Series(price,index=days)
print(s)
s['monday':]
s.iloc[0:3] # index values we use iloc
print(s>6000)
print(s<6000)# it gives the bool values
s[s>6000] # it gives the values
print(s)
```

```python
import pandas as pd
price=[5000,None,6000,9000,None,4000]
s=pd.Series(price)
print(s)
s.isna() # if we use isna it will assign for values(false) and none values (true)
s.isna().sum() # it will count the none values
s.dropna() # it dont show the none values while display the values
s=s.fillna(2000)



```
